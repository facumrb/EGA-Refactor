```python
import ipywidgets as widgets
from IPython.display import display, HTML
import yaml
import os
import sys
```

```python
# Ruta a los archivos
config_path = os.path.join(os.getcwd(), 'config.yaml')
sys.path.append(os.getcwd())

def load_config(path):
    with open(path, 'r') as file:
        return yaml.safe_load(file)

def save_config(path, config_data):
    with open(path, 'w') as file:
        yaml.safe_dump(config_data, file)

# Cargar la configuración inicial
config = load_config(config_path)
```

```python
# Crear widgets interactivos
common_seed_widget = widgets.IntText(
    value=config['common_seed'],
    description='Semilla:',
    disabled=False
)

strategy_widget = widgets.Dropdown(
    options=['uniform', 'center'],
    value=config['ega_params']['strategy'],
    description='Estrategia:',
    disabled=False
)

population_size_widget = widgets.IntSlider(
    value=config['ega_params']['populationSize'],
    min=10,
    max=100,
    step=1,
    description='Tamaño población:',
    disabled=False
)

generations_widget = widgets.IntSlider(
    value=config['ega_params']['generations'],
    min=10,
    max=200,
    step=1,
    description='Generaciones:',
    disabled=False
)

# Widget para ejecutar la simulación
run_button = widgets.Button(
    description='Ejecutar simulación',
    disabled=False,
    button_style='',
    tooltip='Ejecutar EGA con los parámetros configurados'
)

# Botones para ejecutar los scripts de visualización
plot_3d_trajectory_button = widgets.Button(
    description='Visualizar trayectoria 3D',
    disabled=False,
    button_style='',
    tooltip='Ejecutar visualización de trayectoria 3D'
)
plot_best_trajectory_button = widgets.Button(
    description='Visualizar mejor trayectoria',
    disabled=False,
    button_style='',
    tooltip='Ejecutar mejor trayectoria'
)
plot_complexity_vs_fitness_button = widgets.Button(
    description='Visualizar complejidad vs fitness',
    disabled=False,
    button_style='',
    tooltip='Ejecutar visualización de complejidad vs fitness'
)
plot_fitness = widgets.Button(
    description='Visualizar fitness',
    disabled=False,
    button_style='',
    tooltip='Ejecutar visualización de fitness'
)
plot_gene_network_final_state_button = widgets.Button(
    description='Visualizar red genética final',
    disabled=False,
    button_style='',
    tooltip='Ejecutar visualización de red genética final'
)
plot_gene_network_transition_button = widgets.Button(
    description='Visualizar transición de la red genética',
    disabled=False,
    button_style='',
    tooltip='Ejecutar visualización de la transición de la red genética'
)
plot_heatmap_button = widgets.Button(
    description='Visualizar mapa de calor',
    disabled=False,
    button_style='',
    tooltip='Ejecutar visualización de mapa de calor'
)
plot_spaghetti_button = widgets.Button(
    description='Visualizar spaghetti plot',
    disabled=False,
    button_style='',
    tooltip='Ejecutar visualización de spaghetti plot'
)
plot_target_comparison_button = widgets.Button(
    description='Visualizar comparación con el objetivo',
    disabled=False,
    button_style='',
    tooltip='Ejecutar visualización de la comparación al objetivo'
)
```

```python
# Agrupar los botones de visualización en un diccionario
plot_buttons = {
    './plots/plot_3d_trajectory': plot_3d_trajectory_button,
    './plots/plot_best_trajectory': plot_best_trajectory_button,
    './plots/plot_complexity_vs_fitness': plot_complexity_vs_fitness_button,
    './plots/plot_fitness': plot_fitness,
    './plots/plot_gene_network_final_state': plot_gene_network_final_state_button,
    './plots/plot_gene_network_transition': plot_gene_network_transition_button,
    './plots/plot_heatmap': plot_heatmap_button,
    './plots/plot_spaghetti': plot_spaghetti_button,
    './plots/plot_target_comparison': plot_target_comparison_button
}
```

```python
# Función para manejar el botón de ejecución
def on_run_button_clicked(b):
    # Actualizar configuración con los valores de los widgets
    config['common_seed'] = common_seed_widget.value
    config['ega_params']['strategy'] = strategy_widget.value
    config['ega_params']['populationSize'] = population_size_widget.value
    config['ega_params']['generations'] = generations_widget.value
    
    # Guardar configuración actualizada
    save_config(config_path, config)
    
    # Ejecutar run_demo.py
    !python run_demo.py --config config.yaml

def create_plot_callback(script_name):
    def on_button_clicked(b):
        try:
            get_ipython().run_line_magic('run', script_name)
        except Exception as error:
            print(f"Error al ejecutar {script_name}: {error}")
    return on_button_clicked


run_button.on_click(on_run_button_clicked)
for script_path, button in plot_buttons.items():
    full_script_path = os.path.join(os.getcwd(), f'./plots/{script_path}.py')
    button.on_click(create_plot_callback(full_script_path))
```
